```python
class Chocolate:
    def __init__(self, price: float, additive: str = None):
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля")
        self.price = price
        self.additive = additive

    def show_my_chocolate(self) -> None:
        if self.additive:
            print(f"Шоколадный батончик, добавка: {self.additive}")
        else:
            print("Обычный шоколадный батончик")

    def __str__(self) -> str:
        return f"Шоколадный батончик, {self.price} рублей"


if __name__ == "__main__":
    chocolate1 = Chocolate(59.9, "орехи")
    chocolate2 = Chocolate(45.0)
    chocolate3 = Chocolate(120.5, "карамель и соль")

    print("--- Вызов show_my_chocolate ---")
    chocolate1.show_my_chocolate()
    chocolate2.show_my_chocolate()
    chocolate3.show_my_chocolate()

    print("\n--- Вывод через print() ---")
    print(chocolate1)
    print(chocolate2)
    print(chocolate3)

    try:
        bad_chocolate = Chocolate(-10.0)
    except ValueError as e:
        print(f"\nОшибка: {e}")
```
