```python
import pandas as pd
import numpy as np

df = pd.read_csv('train.csv')

print("Информация о типах данных:")
print(df.info())
print("\nКоличество пропусков:")
print(df.isnull().sum())
print("\nСтатистика числовых признаков:")
print(df.describe())

survival_by_class = df.groupby('Pclass')['Survived'].mean() * 100
print("\nПроцент выживаемости по классам:")
print(survival_by_class)

def extract_first_name(full_name):
    parts = full_name.split(',')
    if len(parts) > 1:
        after_comma = parts[1].strip()
        if '(' in after_comma and ')' in after_comma:
            start = after_comma.index('(') + 1
            end = after_comma.index(')')
            name_in_brackets = after_comma[start:end].strip()
            first_name = name_in_brackets.split()[0] if name_in_brackets else ''
        else:
            if '.' in after_comma:
                name_part = after_comma.split('.', 1)[1].strip()
                first_name = name_part.split()[0] if name_part else ''
            else:
                first_name = after_comma.split()[0] if after_comma else ''
    else:
        first_name = full_name.split()[0] if full_name else ''
    return first_name

df['First_Name'] = df['Name'].apply(extract_first_name)

male_names = df[df['Sex'] == 'male']['First_Name']
female_names = df[df['Sex'] == 'female']['First_Name']

most_common_male = male_names.value_counts().idxmax()
most_common_female = female_names.value_counts().idxmax()
print(f"\nСамое популярное мужское имя: {most_common_male}")
print(f"Самое популярное женское имя: {most_common_female}")

print("\nСамое популярное имя по классам и полу:")
for pclass in df['Pclass'].unique():
    print(f"Класс {pclass}:")
    class_df = df[df['Pclass'] == pclass]
    male_names_class = class_df[class_df['Sex'] == 'male']['First_Name']
    female_names_class = class_df[class_df['Sex'] == 'female']['First_Name']
    if not male_names_class.empty:
        most_male = male_names_class.value_counts().idxmax()
        print(f"  Мужское: {most_male}")
    else:
        print("  Мужское: нет данных")
    if not female_names_class.empty:
        most_female = female_names_class.value_counts().idxmax()
        print(f"  Женское: {most_female}")
    else:
        print("  Женское: нет данных")

print("\nПассажиры старше 44 лет:")
print(df[df['Age'] > 44].head(10))

print("\nПассажиры мужского пола младше 44 лет:")
print(df[(df['Age'] < 44) & (df['Sex'] == 'male')].head(10))

cabin_group_sizes = df.groupby('Cabin').size()
size_distribution = cabin_group_sizes.value_counts().sort_index()
print("\nКоличество кают с определённым числом пассажиров:")
print(size_distribution)
```
