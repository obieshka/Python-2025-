```python
import requests
import pandas as pd

def get_all_books():
    base_url = "https://www.anapioficeandfire.com/api/books"
    books = []
    page = 1
    while True:
        response = requests.get(base_url, params={"page": page, "pageSize": 50})
        if response.status_code != 200:
            break
        data = response.json()
        if not data:
            break
        books.extend(data)
        page += 1
    return pd.DataFrame(books)

df_books = get_all_books()
print(df_books.shape)
df_books.head()
```

```python
def get_all_houses():
    base_url = "https://www.anapioficeandfire.com/api/houses"
    houses = []
    page = 1
    while True:
        response = requests.get(base_url, params={"page": page, "pageSize": 50})
        if response.status_code != 200:
            break
        data = response.json()
        if not data:
            break
        houses.extend(data)
        page += 1
    return pd.DataFrame(houses)

df_houses = get_all_houses()
print(df_houses.shape)
df_houses.head()
```

```python
def get_houses_with_titles():
    base_url = "https://www.anapioficeandfire.com/api/houses"
    houses = []
    page = 1
    while True:
        response = requests.get(base_url, params={"page": page, "pageSize": 50, "hasTitles": True})
        if response.status_code != 200:
            break
        data = response.json()
        if not data:
            break
        houses.extend(data)
        page += 1
    return pd.DataFrame(houses)

df_houses_titles = get_houses_with_titles()
print(df_houses_titles.shape)
df_houses_titles.head()
```
